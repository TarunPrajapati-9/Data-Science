```python
pd.get_dummies(en_data).info()
```

```python
ohe = OneHotEncoder() 
# ohe = OneHotEncoder(drop="first") drop first column of two categorical data like Gender_Male and Gender_Female it drops Gender_male
ar  = ohe.fit_transform(en_data).toarray()
```
